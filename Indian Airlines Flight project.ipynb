##  Creating a Backup of Original Dataset

Creating a copy of the original dataset to preserve raw data before performing data cleaning and preprocessing operations.

```python
df_clean = df.copy()
```

##  Standardizing Column Names

Removing leading and trailing spaces from column names to maintain consistency.

```python
df_clean.columns = df_clean.columns.str.strip()
```

##  Standardizing Text Data

Removing unnecessary leading and trailing spaces from categorical text columns to maintain data consistency

```python
for col in df_clean.select_dtypes(include='object').columns:
    df_clean[col] = df_clean[col].str.strip()
```

##  Date Format Conversion

Converting the flight date column into datetime format to support date-based analysis.

```python
df_clean['Date'] = pd.to_datetime(
    df_clean['Date'],
    errors='coerce'
)
```

```python
df_clean['Date'].dtype
```

## Checking Negative Values

Identifying negative values in numerical columns to detect potentially invalid records and data entry errors.

```python
numeric_cols = df_clean.select_dtypes(
    include=np.number
).columns

(df_clean[numeric_cols] < 0).sum()
```

##  Outlier Detection Using IQR Method

Using the Interquartile Range (IQR) method to identify potential outliers in numerical columns.

```python
Q1 = df_clean['Arrival_Delay_Min'].quantile(0.25)

Q3 = df_clean['Arrival_Delay_Min'].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = df_clean[
    (df_clean['Arrival_Delay_Min'] < lower_limit) |
    (df_clean['Arrival_Delay_Min'] > upper_limit)
]

print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
print("Total Outliers:", len(outliers))
```

# Exploratory Data Analysis (EDA)

Exploratory Data Analysis is performed to understand the patterns, 
relationships, and distributions within the airline dataset.

Various data visualization techniques are used to analyze flight 
delays, revenue, passenger trends, airline performance, and 
operational efficiency.

##  Setting Airline Visualization Theme

Defining a consistent airline-inspired color palette and visualization 
style for all charts to maintain a professional appearance.

```python
sns.set_theme(style="whitegrid")

flight_colors = [
    "#271D5C",  # dark Blue
    "#00A6D6",  # Sky Blue
    "#FC6A1C",  # orange
    "#DE1F1F",  # Red
    "#87CEEB"   # Light Blue
]
```

##  Scatter Plot – Passengers vs Revenue

A scatter plot is used to analyze the relationship between the number 
of passengers and revenue generated by flights.

This visualization helps identify patterns and correlations between 
passenger traffic and airline revenue.

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df_clean,
    x='Passengers',
    y='Revenue_INR',
    hue='Airline',
    palette=flight_colors,
    alpha=0.7
)

plt.title('Passengers vs Revenue Analysis', fontsize=15)
plt.xlabel('Number of Passengers')
plt.ylabel('Revenue (INR)')
plt.legend(title='Airline')
plt.show()
```

##  Histogram – Arrival Delay Distribution

A histogram is used to understand the frequency distribution of 
arrival delays in minutes.

It helps identify common delay patterns and extreme delay values.

```python
plt.figure(figsize=(10, 6))

sns.histplot(
    data=df_clean,
    x='Arrival_Delay_Min',
    bins=30,
    kde=True,
    color='#8D0FBF'
)

plt.title('Distribution of Flight Arrival Delays', fontsize=15)
plt.xlabel('Arrival Delay (Minutes)')
plt.ylabel('Frequency')
plt.show()
```

##  Bar Chart – Average Delay by Airline

A bar chart is used to compare the average arrival delay of different 
airlines.

This analysis helps identify airlines with higher and lower average 
flight delays.

```python
airline_delay = (
    df_clean.groupby('Airline')['Arrival_Delay_Min']
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(6, 6))

plt.bar(
    airline_delay.index,
    airline_delay.values,
    color=flight_colors
)

plt.title('Average Arrival Delay by Airline', fontsize=15)
plt.xlabel('Airline')
plt.ylabel('Average Arrival Delay (Minutes)')
plt.xticks(rotation=30)
plt.show()
```

##  Pairplot – Flight Data Relationships

A pairplot is used to visualize pairwise relationships between 
selected numerical variables.

It helps understand the relationship between passengers, revenue, 
flight duration, and arrival delays.

```python
pair_data = df_clean[
    [
        'Passengers',
        'Revenue_INR',
        'Flight_Duration_Min',
        'Arrival_Delay_Min'
    ]
].dropna()

pair_data = pair_data.sample(
    n=min(1000, len(pair_data)),
    random_state=42
)

sns.pairplot(
    pair_data,
    diag_kind='hist',
    plot_kws={'color': '#DE1F1F', 'alpha': 0.5},
    diag_kws={'color': '#E3DD0E'}
)

plt.show()
```

##  Violin Plot – Arrival Delay by Airline

A violin plot is used to visualize the distribution and density 
of arrival delays across different airlines.

It helps compare delay patterns and variations between airlines.

```python
plt.figure(figsize=(11, 6))

sns.violinplot(
    data=df_clean,
    x='Airline',
    y='Arrival_Delay_Min',
    palette=flight_colors,
    inner='quartile'
)

plt.title('Arrival Delay Distribution by Airline', fontsize=15)
plt.xlabel('Airline')
plt.ylabel('Arrival Delay (Minutes)')
plt.xticks(rotation=30)
plt.show()
```

##  Correlation Heatmap – Flight Data

A correlation heatmap is used to identify relationships between 
numerical variables in the airline dataset.

It helps understand how flight duration, distance, passengers, 
revenue, operating cost, and delays are related.

```python
plt.figure(figsize=(14, 9))

corr = df_clean.select_dtypes(
    include=np.number
).corr()

sns.heatmap(
    corr,
    annot=True,
    cmap='YlGnBu',
    fmt='.2f',
    linewidths=0.5
)

plt.title('Correlation Heatmap - Indian Airlines', fontsize=16)
plt.show()
```

##  Pie Chart – Flight Distribution by Airline

A pie chart is used to visualize the percentage distribution of 
flights operated by different airlines.

It helps identify the contribution of each airline to the 
total number of flights.

```python
airline_flights = df_clean['Airline'].value_counts()

plt.figure(figsize=(9, 7))

plt.pie(
    airline_flights,
    labels=airline_flights.index,
    autopct='%1.1f%%',
    colors=flight_colors,
    startangle=90,
    explode=[0.03] * len(airline_flights)
)

plt.title('Flight Distribution by Airline', fontsize=16)

plt.axis('equal')
plt.show()
```

##  Line Chart – Monthly Revenue Trend

A line chart is used to analyze monthly revenue trends over time.

It helps identify revenue growth, seasonal patterns, and 
changes in airline business performance.

```python
monthly_revenue = (
    df_clean
    .groupby(df_clean['Date'].dt.to_period('M'))['Revenue_INR']
    .sum()
)

plt.figure(figsize=(12, 6))

plt.plot(
    monthly_revenue.index.astype(str),
    monthly_revenue.values,
    color='#1D24E0',
    marker='o',
    linewidth=2,
    markersize=5
)

plt.title('Monthly Airline Revenue Trend', fontsize=16)
plt.xlabel('Month')
plt.ylabel('Total Revenue (INR)')

plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

##  KDE Plot – Arrival Delay Distribution

A KDE (Kernel Density Estimation) plot is used to visualize the 
probability density distribution of flight arrival delays.

It helps understand the concentration and spread of arrival delays.

```python
plt.figure(figsize=(10, 6))

sns.kdeplot(
    data=df_clean,
    x='Arrival_Delay_Min',
    fill=True,
    color='#E01DDA',
    linewidth=2
)

plt.title('KDE Plot - Arrival Delay Distribution', fontsize=15)
plt.xlabel('Arrival Delay (Minutes)')
plt.ylabel('Density')

plt.show()
```

## Count Plot – Flight Count by Airline

A count plot is used to display the total number of flights 
operated by each airline.

It helps compare the operational activity of different airlines.

```python
plt.figure(figsize=(8, 6))

sns.countplot(
    data=df_clean,
    x='Airline',
    hue='Airline',
    palette=flight_colors,
    legend=False
)

plt.title('Total Flights by Airline', fontsize=15)
plt.xlabel('Airline')
plt.ylabel('Number of Flights')
plt.xticks(rotation=30)

plt.show()
```

##   Cat Plot – Airline Delay Comparison

A categorical plot is used to compare arrival delays across 
different airlines.

It helps analyze airline-wise flight delay patterns.

```python
sns.catplot(
    data=df_clean,
    x='Airline',
    y='Arrival_Delay_Min',
    kind='box',
    palette=flight_colors,
    height=6,
    aspect=1.5
)

plt.title('Airline-wise Arrival Delay Comparison')
plt.xlabel('Airline')
plt.ylabel('Arrival Delay (Minutes)')
plt.xticks(rotation=30)

plt.show()
```

##  Cluster Map – Numerical Flight Data

A cluster map is used to visualize relationships between 
numerical flight variables through hierarchical clustering.

It helps identify groups of variables with similar 
correlation patterns.

```python
cluster_data = df_clean[
    [
        'Passengers',
        'Distance_km',
        'Flight_Duration_Min',
        'Departure_Delay_Min',
        'Arrival_Delay_Min',
        'Revenue_INR',
        'Operating_Cost_INR_Est',
        'Estimated_Profit_INR',
        'Customer_Rating'
    ]
].corr()

sns.clustermap(
    cluster_data,
    annot=True,
    cmap='YlGnBu',
    fmt='.2f',
    figsize=(12, 12),
    linewidths=0.5
)

plt.show()
```
